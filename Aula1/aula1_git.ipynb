```python
# Helper multiplataforma: substitui o `ls` do Linux/Mac, que não existe no Windows
from pathlib import Path

def listar(pasta='.', ocultos=False):
    """Lista o conteúdo de uma pasta. Funciona igual no Windows, macOS e Linux."""
    itens = sorted(
        p.name + ('/' if p.is_dir() else '')
        for p in Path(pasta).iterdir()
        if ocultos or not p.name.startswith('.')
    )
    print('  '.join(itens) if itens else '(pasta vazia)')
```

## Parte 4 — Repositório remoto (GitHub)

> **Antes de começar: o GitHub não aceita a senha da sua conta no terminal.**
> Quando o Git pedir `Password`, ele espera um **token de acesso**. Gere o seu agora:
>
> 1. No GitHub, clique na sua foto (canto superior direito) → **Settings**
> 2. Role a barra lateral esquerda até o fim → **Developer settings**
> 3. **Personal access tokens** → **Tokens (classic)** → **Generate new token (classic)**
> 4. *Note:* `curso-ada` · *Expiration:* 90 dias · *Scopes:* marque a caixa **`repo`**
> 5. **Generate token** e **copie o token imediatamente** — ele começa com `ghp_` e o GitHub nunca mais vai mostrá-lo
>
> Na hora do `push`, informe seu **usuário do GitHub** e cole o **token** no lugar da senha.
> Ao colar no terminal, **nada aparece na tela** — nem asteriscos. Isso é normal: cole e aperte Enter.
> Depois disso o Git guarda a credencial e não pergunta mais. (No Linux, rode antes `git config --global credential.helper store`.)

### Exercício 4.1
Conecte seu repositório local ao remoto do GitHub e envie a branch `main`, configurando o rastreamento (*upstream*).

> ⚠️ **`git remote add` só funciona uma vez.** Se você rodar esta célula de novo, verá `error: remote origin already exists` — o Git recusa criar um segundo remoto com o mesmo nome e **mantém o antigo**. Se a URL tiver mudado, o push continua indo para o endereço velho.
>
> Por isso a célula abaixo verifica antes: usa `add` na primeira vez e `set-url` (que **troca** o endereço) nas seguintes.

```python
# Sua solução aqui
!git remote add origin https://github.com/danielmassita/meu-primeiro-repo.git
!git branch -M main
!git push -u origin main

URL = "https://github.com/danielmassita/meu-primeiro-repo.git"   # troque SEU-USUARIO

import subprocess
ja_existe = 'origin' in subprocess.run(['git', 'remote'],
                                       capture_output=True, text=True).stdout.split()
if ja_existe:
    !git remote set-url origin {URL}    # já havia um remoto: troca a URL
else:
    !git remote add origin {URL}        # primeira vez: cria o remoto

!git remote -v
!git push -u origin main
```

### Gabarito comentado

1. **Working directory** é a pasta onde você edita os arquivos. **Staging area** é a área intermediária onde você seleciona o que vai entrar no próximo commit (`git add`). **Repositório local** é o histórico de commits já gravado no `.git` (`git commit`).
2. `git fetch` baixa as alterações do remoto mas **não** as aplica na sua branch — você fica sabendo que existem. `git pull` é um `fetch` seguido de um `merge` automático na sua branch atual.
3. O Git não consegue decidir sozinho e gera um **conflito de merge**: ele marca o trecho no arquivo com `<<<<<<<`, `=======`, `>>>>>>>` e cabe a você editar, escolher a versão final, dar `add` e commitar.
4. Dados grandes incham o repositório de forma permanente (o histórico guarda todas as versões) e deixam clones lentos; credenciais commitadas ficam no histórico mesmo depois de removidas do arquivo, e devem ser consideradas comprometidas. Use `.gitignore` e variáveis de ambiente.
5. Porque análise é um processo iterativo e experimental — o versionamento permite reproduzir um resultado, comparar abordagens, colaborar sem sobrescrever o trabalho alheio e construir um portfólio público.
